# 01 - Chamada simples

Primeira chamada à API da OpenAI usando o pacote oficial `openai`.

A chave da API e o modelo são lidos do arquivo `.env` na raiz do projeto:

```env
OPENAI_API_KEY=...
OPENAI_MODEL=gpt-5-mini
```

````python
import os

from dotenv import find_dotenv, load_dotenv
from IPython.display import Markdown, display
from openai import OpenAI

load_dotenv(find_dotenv())
````

````python
client = OpenAI(api_key=os.getenv("OPENAI_API_KEY"))
modelo = os.getenv("OPENAI_MODEL", "gpt-5-mini")
````

````python
pergunta = "Quem foi Isaac Newton?"

resposta = client.responses.create(
    model=modelo,
    input=pergunta,
)
````

````python
display(Markdown(f"**Pergunta:** {pergunta}\n\n**Resposta:**\n\n{resposta.output_text}"))
````
