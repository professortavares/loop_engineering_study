# 02 - Chamada simples com LangChain

Mesma chamada do notebook `01-Chamada simples - OpenAI`, agora usando o framework **LangChain** com a integração `langchain-openai`.

A chave da API e o modelo continuam vindo do arquivo `.env` na raiz do projeto.

```python
import os

from dotenv import find_dotenv, load_dotenv
from IPython.display import Markdown, display
from langchain_openai import ChatOpenAI

load_dotenv(find_dotenv())

llm = ChatOpenAI(
    model=os.getenv("OPENAI_MODEL", "gpt-5-mini"),
    api_key=os.getenv("OPENAI_API_KEY"),
)
```

```python
pergunta = "Quem foi Isaac Newton?"

resposta = llm.invoke(pergunta)
```

```python
display(Markdown(f"**Pergunta:** {pergunta}\n\n**Resposta:**\n\n{resposta.content}"))
```

## Vantagens de usar o LangChain

- **Interface única para modelos:** trocar de OpenAI para Anthropic, Google, Ollama etc. muda só a classe do modelo; o restante do código fica igual.
- **Composição com cadeias (LCEL):** prompts, modelos e parsers são encadeados com o operador `|`, em vez de montar cada etapa à mão.
- **Templates de prompt reutilizáveis:** variáveis e instruções ficam separadas do código de chamada.
- **Saída estruturada e parsers:** converte a resposta em texto, JSON ou objetos tipados (por exemplo, com Pydantic).
- **Ferramentas e agentes:** integra chamadas de função, buscas e fluxos com múltiplos passos, como os que o LangGraph orquestra.
- **Memória e histórico de conversa:** suporte pronto para manter contexto entre chamadas.
- **RAG pronto:** carregadores de documentos, divisores de texto, embeddings e bancos vetoriais já integrados.
- **Observabilidade:** integra com ferramentas de rastreio, como o LangSmith, para depurar cada etapa da cadeia.
- **Ecossistema amplo:** muitas integrações mantidas pela comunidade e pela equipe do framework.
