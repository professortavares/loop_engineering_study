# 01 - Deep Agents

Um **deep agent** é um agente pronto, criado com `create_deep_agent`, feito para tarefas complexas. Ele é um grafo do LangGraph, então o `invoke` funciona como nos notebooks anteriores.

Neste primeiro notebook, o agente só responde a uma pergunta simples, sem ferramentas. Os próximos notebooks vão acrescentar recursos, como planejamento, ferramentas e subagentes.

## 1. Configuração

A chave e o modelo vêm do `.env` na raiz do projeto, como nos outros notebooks.

```python
import os

from dotenv import find_dotenv, load_dotenv
from langchain_core.messages import HumanMessage
from langchain_openai import ChatOpenAI

from deepagents import create_deep_agent

load_dotenv(find_dotenv())

llm = ChatOpenAI(
    model=os.getenv("OPENAI_MODEL", "gpt-5-mini"),
    api_key=os.getenv("OPENAI_API_KEY"),
)
```

## 2. Criando o agente

`create_deep_agent` recebe o modelo e um `system_prompt`, que define o papel do agente.

```python
agente = create_deep_agent(
    model=llm,
    system_prompt="Você é um assistente prestativo. Responda em português, de forma curta e clara.",
)
```

## 3. Fazendo uma pergunta

O `invoke` recebe a lista de mensagens e devolve o estado final. A última mensagem é a resposta do agente.

```python
resultado = agente.invoke({"messages": [HumanMessage(content="Explique em duas frases o que é um agente de IA.")]})

resposta = resultado["messages"][-1]
print(resposta.content)
```

## Observações

- O deep agent é um grafo, então `invoke`, `stream` e o estado funcionam como nos notebooks do LangGraph.
- Sem ferramentas, o agente só usa o modelo para responder.
